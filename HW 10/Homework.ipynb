```python
import cv2
import os

video_path = 'data/video.mp4'
cap = cv2.VideoCapture(video_path)

ret, frame = cap.read()
if not ret:
    print("Unable to open a video")
    exit()

print("Choose what to track and press ENTER")
bbox = cv2.selectROI("Select Object", frame, fromCenter=False, showCrosshair=True)
cv2.destroyWindow("Select Object")

def run_tracker(tracker_type, start_bbox, video_cap, output_folder):
    if tracker_type == "MIL":
        tracker = cv2.TrackerMIL_create()
    elif tracker_type == "CSRT":
        tracker = cv2.TrackerCSRT_create()
    
    video_cap.set(cv2.CAP_PROP_POS_FRAMES, 0)
    success, frame = video_cap.read()
    tracker.init(frame, start_bbox)
    
    print(f"Tracker start {tracker_type}...")

    for i in range(100):
        success, frame = video_cap.read()
        if not success:
            break

        success, box = tracker.update(frame)

        if success:
            (x, y, w, h) = [int(v) for v in box]
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
            cv2.putText(frame, f"{tracker_type} Frame {i}", (20, 40), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
        
        cv2.imshow("Tracking Process", frame)
        
        if cv2.waitKey(15) & 0xFF == ord('q'):
            break
    cv2.destroyAllWindows()

run_tracker("CSRT", bbox, cap, "output_mil")
run_tracker("MIL", bbox, cap, "output_kcf")

cap.release()
```
