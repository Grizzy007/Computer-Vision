```python
# Importing neccessary libraries
import cv2
import numpy as np
from matplotlib import pyplot as plt
import dlib
```

```python
# Reading image and convert color to RGB anf Gray
img = cv2.imread('data/team.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(img)
```

```python
# Initializing face detector
detector = dlib.get_frontal_face_detector()
# Using face detector to detect faces on the image
rects = detector(gray, 1)

print(rects)
```

```python
# Converting rectengle cordinates from detector to fit the image
def rects_coordinate(rect):
    x=rect.left()
    y=rect.top()
    w=rect.right() - x
    h=rect.bottom() - y

    return (x, y, w, h)
```

```python
# Defining function to draw rectangles on the image
def print_rects(img, rects):
    result = np.copy(img)

    for rect in rects:
        # Get coresponding cordinates from detector
        x, y, w, h = rects_coordinate(rect)
        # Printing rectangles on the image
        cv2.rectangle(result, (x, y), (x+w, y+h), (0+x, 255-x, 0+y), 3)
        
    return result
```

```python
plt.imshow(print_rects(img, rects))
```

```python
# Using more difficult image
image = cv2.imread('data/Mbappe-PSG.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
```
